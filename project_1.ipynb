```python
import pandas as pd
import numpy as np
df = pd.read_csv('SampleSuperstore.csv')
```

```python
df['Profit_Margin'] = df['Profit'] / df['Sales'].replace(0, np.nan)
df.head()
```

```python
Cat_sub = df.groupby(['Category','Sub-Category']).agg(
    Total_Sales = ('Sales','sum'),
    Total_Profit = ('Profit','sum'),
    Avg_Discount = ('Discount','mean'),
    Avg_Margin = ('Profit_Margin','mean')
).sort_values('Total_Profit')
# df.groupby(['Category', 'Sub-Category']).agg({'Sales': 'sum', 'Profit':'sum', 'Discount': 'mean', 'Profit Margin': 'mean'}).sort_values(by='Profit')
Cat_sub
```

```python
sort = df.groupby(['Sub-Category']).sum(numeric_only = True).idxmin()['Profit']
sort
print(f'The Sub-Category that loses the most money is {sort}')

Table_loss_percent = Cat_sub.loc[('Furniture', 'Tables'), 'Avg_Discount']*100
Bookcases_loss_percent = Cat_sub.loc[('Furniture', 'Bookcases'), 'Avg_Discount']*100

furn_loss = Cat_sub.loc[('Furniture',['Tables','Bookcases']), 'Total_Profit'].sum()*(-1)
 
```

```python
bins = [-0.001, 0, 0.2, 0.5, 1.0] 
labels = ['No Discount', 'Low', 'Medium', 'High']
df['Discount_Bin'] = pd.cut(df['Discount'], bins=bins, labels=labels, include_lowest=True)
#df['Discount_Bin'] = df['Discount'].apply(lambda x: 'No Discount' if x == 0 else 'Low' if x <= 0.2 else 'Medium' if x <= 0.5 else 'High Discount')
df.head()
```

```python
# High_Discount_Profit = df.groupby('Discount_Bin').agg({"Quantity": "count", "Sales": "sum", 'Profit': "sum"}).min()['Profit']
Discounts = df.groupby('Discount_Bin').agg(
    Total_Sales = ('Sales','sum'),
    Total_Profit = ('Profit','sum'),
    Order_count = ('Quantity', 'count'),
    Avg_Profit_Margin = ('Profit_Margin', 'mean')
)

```

```python
High_Discount_Profit = Discounts.loc['High', 'Total_Profit']*(-1)
print(f"High Discount has the highest amount of negative profits {High_Discount_Profit}" )
```

```python
df.groupby(['Region', 'Segment']).agg(Total_sales = ('Sales', 'sum'), Total_profit = ('Profit', 'sum'))
```

```python
most_profitable = df.groupby(['Region', 'Segment']).agg({'Sales': 'sum', "Profit": 'sum'}).idxmax()['Profit']
least_profitable = df.groupby(['Region', 'Segment']).agg({'Sales': 'sum', "Profit": 'sum'}).idxmin()['Profit']

print(f"The Region-Segment combination is most profitable in {most_profitable} and least profitable in {least_profitable}")

region_profit = df.groupby('Region')['Profit'].sum()
loss_regions = region_profit[region_profit < 0]
print('Regions with overall loss:', loss_regions.index.tolist()) 
```

```python

Neg_ation = df[df['Profit'] < 0]
Neg_ation.groupby(['Sub-Category','Region','Segment'])['Profit'].sum().sort_values().head(3)
```

```python
Unprofitable_orders = Neg_ation[['Sub-Category', 'Region','Segment']].value_counts().idxmax()

print(f"The Sub-Category, Region, and Segment  that appear most frequently in unprofitable orders are {Unprofitable_orders} with a loss of $13,407")
```

```python
Percentage = Neg_ation['Discount_Bin'].value_counts(normalize=True) * 100
Percentage
```

```python
Percent = Percentage.loc[('High')] + Percentage.loc[('Medium')]
```

```python
print(f"""The company is bleeding money primarily through aggressive discounting on a few problem products.

       Tables and Bookcases (Furniture) are sold at average discounts of {Table_loss_percent}% and {Bookcases_loss_percent}%, leading to combined losses of over ${furn_loss}K. 

       In Office Supplies, Binders are heavily discounted (avg 37%) and lose money, especially in the Central region among Consumer customers. 

       Meanwhile, high-discount (~{Percentage.loc[('High')]}%) sales – just 856 transactions – account for a staggering -${High_Discount_Profit}K in profit. 

       My recommendation: review discount policies for Tables and Binders immediately, and implement a discount cap or approval workflow for discounts above 40%.""")
```
